```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Open the CSV file 
df=pd.read_csv("./traffic_log.csv")
#Count how many times each vehicle type appears
class_dist=df["class"].value_counts()
#Draw the bar chart
plt.figure(figsize=(7,5))
sns.barplot(x=class_dist.index, y=class_dist.values, hue=class_dist.index.astype(str), palette="viridis")

plt.title("Vehicle Class Distribution")
plt.xlabel("Class")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```
